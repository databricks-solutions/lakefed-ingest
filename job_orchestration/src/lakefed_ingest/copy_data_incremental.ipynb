```sql
DECLARE OR REPLACE qry_str STRING;

-- Build temp view query string.
-- When use_remote_query is true, pushes the SELECT + watermark WHERE clause to the source
-- via remote_query (required when no Unity Catalog foreign catalog is configured).
-- Otherwise uses Lakehouse Federation (default).
SET VAR qry_str = CASE WHEN :use_remote_query = 'true' THEN
    'CREATE OR REPLACE TEMP VIEW vw_src AS '
    || 'SELECT * FROM remote_query('
    || chr(39) || :src_connection || chr(39)
    || CASE :src_type
         WHEN 'oracle'  THEN ', service_name => ' || chr(39) || :src_database || chr(39)
         WHEN 'db2_luw' THEN ''  -- database is embedded in the JDBC URL; cannot be passed externally
         ELSE                ', database => '     || chr(39) || :src_database || chr(39)
       END
    || ', query => ' || chr(39)
    ||   'SELECT ' || :select_list || ' FROM ' || :src_schema || '.' || :src_table
    || CASE WHEN :watermark_col_type = 'timestamp'
         THEN ' WHERE ' || :watermark_col_name || ' > '  || chr(92)||chr(39) || :watermark_col_start_value || chr(92)||chr(39)
           || ' AND '  || :watermark_col_name || ' <= ' || chr(92)||chr(39) || :watermark_col_end_value   || chr(92)||chr(39)
         ELSE ' WHERE ' || :watermark_col_name || ' > '  || :watermark_col_start_value
           || ' AND '  || :watermark_col_name || ' <= ' || :watermark_col_end_value
         END
    || chr(39) || ')'
  ELSE
    'CREATE OR REPLACE TEMP VIEW vw_src AS '
    || 'SELECT ' || :select_list || ' FROM ' || :src_catalog || '.' || :src_schema || '.' || :src_table
    || ' WITH(fetchSize = 200000)'
    || CASE WHEN :watermark_col_type = 'timestamp'
         THEN ' WHERE ' || :watermark_col_name || ' > '  || chr(39) || :watermark_col_start_value || chr(39)
           || ' AND ' || :watermark_col_name || ' <= ' || chr(39) || :watermark_col_end_value || chr(39)
         ELSE ' WHERE ' || :watermark_col_name || ' > '  || :watermark_col_start_value
           || ' AND ' || :watermark_col_name || ' <= ' || :watermark_col_end_value
         END
  END;

EXECUTE IMMEDIATE qry_str;
SELECT qry_str;
```

```sql
create or replace temporary function get_join_clause(primary_key string)
  returns string
  deterministic
  comment 'Converts a comma-separated list of primary key columns into a join clause'
  language python
  as $$
    pk_list = [col.strip() for col in primary_key.split(',')]
    join_clause = ' and '.join([f't.{col} = s.{col}' for col in pk_list])
    return join_clause
  $$;
```

```sql
begin
  declare qry_str string;
  declare join_clause string;
  
  -- Get join clause
  set join_clause = (select get_join_clause(:primary_key));
  
  -- Build merge statement
  set qry_str = 
    "merge into " || :sink_catalog || '.' || :sink_schema || '.' || :sink_table || " as t " ||
    "using vw_src as s " ||
    "on " || join_clause || " " ||
    "when matched then update set * " ||
    "when not matched then insert *";
    
  -- Execute merge statement
  execute immediate qry_str;
  select qry_str;
end
```
